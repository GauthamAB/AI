```python
import heapq


def heuristic(state, goal):
    return sum(
        1 for i in range(9)
        if state[i] != 0 and state[i] != goal[i]
    )


def astar(start, goal):

    pq = [(0, start, [])]
    visited = set()

    while pq:

        cost, state, path = heapq.heappop(pq)

        if state == goal:
            return path + [state]

        if state in visited:
            continue

        visited.add(state)

        zero = state.index(0)

        moves = []

        if zero >= 3:
            moves.append(zero - 3)

        if zero < 6:
            moves.append(zero + 3)

        if zero % 3 != 0:
            moves.append(zero - 1)

        if zero % 3 != 2:
            moves.append(zero + 1)

        for move in moves:

            new_state = list(state)

            new_state[zero], new_state[move] = (
                new_state[move], new_state[zero]
            )

            new_state = tuple(new_state)

            if new_state not in visited:

                g = len(path) + 1
                h = heuristic(new_state, goal)

                f = g + h

                heapq.heappush(
                    pq, (f, new_state, path + [state])
                )

    return None


start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

result = astar(start, goal)

if result:
    for state in result:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()
else:
    print("No solution found")
```
