```python
def dfs(start, goal):
    stack = [(start, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path + [state]

        if state in visited:
            continue

        visited.add(state)

        zero = state.index(0)

        moves = []

        if zero >= 3:
            moves.append(zero - 3)

        if zero < 6:
            moves.append(zero + 3)

        if zero % 3 != 0:
            moves.append(zero - 1)

        if zero % 3 != 2:
            moves.append(zero + 1)

        for move in moves:
            new_state = list(state)

            new_state[zero], new_state[move] = (
                new_state[move], new_state[zero]
            )

            stack.append((tuple(new_state), path + [state]))

    return None


start = (5, 4, 0,
         6, 1, 8,
         7, 3, 2)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

result = dfs(start, goal)

if result:
    for state in result:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()
else:
    print("No solution found")
```
