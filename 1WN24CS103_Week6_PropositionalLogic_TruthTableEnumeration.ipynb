```python
from itertools import product


def propositional_logic():

    print("P Q R | P<->Q | Q OR R | NOT P -> R | KB")
    print("---------------------------------------------")

    models = []

    for P, Q, R in product([True, False], repeat=3):

        s1 = (P == Q)
        s2 = Q or R
        s3 = P or R

        KB = s1 and s2 and s3

        print(P, Q, R, "|", s1, "|", s2, "|", s3, "|", KB)

        if KB:
            models.append((P, Q, R))

    print("\nModels of KB:", models)

    q_entails = all(Q for P, Q, R in models)

    pq_entails = all((not P) or Q for P, Q, R in models)

    rq_entails = all((not R) or Q for P, Q, R in models)

    print("\nKB entails Q:", q_entails)
    print("KB entails P -> Q:", pq_entails)
    print("KB entails R -> Q:", rq_entails)


propositional_logic()
```
