```python
def depth_limited_search(start, goal, limit):

    stack = [(start, 0, [start])]

    while stack:
        state, depth, path = stack.pop()

        if state == goal:
            return path

        if depth < limit:

            zero = state.index(0)

            moves = []

            if zero >= 3:
                moves.append(zero - 3)

            if zero < 6:
                moves.append(zero + 3)

            if zero % 3 != 0:
                moves.append(zero - 1)

            if zero % 3 != 2:
                moves.append(zero + 1)

            for move in moves:
                new_state = list(state)

                new_state[zero], new_state[move] = (
                    new_state[move], new_state[zero]
                )

                new_state = tuple(new_state)

                if new_state not in path:
                    stack.append(
                        (new_state, depth + 1, path + [new_state])
                    )

    return None


def ids(start, goal):

    for depth in range(50):

        result = depth_limited_search(start, goal, depth)

        if result:
            return result

    return None


start = (5, 4, 0,
         6, 1, 8,
         7, 3, 2)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

result = ids(start, goal)

if result:
    for state in result:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()
else:
    print("No solution found")
```
