```python
def print_board(board):
    for i in range(0, 9, 3):
        print(board[i], "|", board[i+1], "|", board[i+2])
        if i < 6:
            print("---------")


def check_winner(board, player):
    win = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]

    for w in win:
        if all(board[i] == player for i in w):
            return True
    return False


def tic_tac_toe():
    board = [" "] * 9

    while True:
        print_board(board)

        pos = int(input("Enter position (1-9): ")) - 1

        if pos < 0 or pos > 8 or board[pos] != " ":
            print("Invalid move")
            continue

        board[pos] = "X"

        if check_winner(board, "X"):
            print_board(board)
            print("Human Wins!")
            break

        if " " not in board:
            print("Draw!")
            break

        # Computer move
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                break

        if check_winner(board, "O"):
            print_board(board)
            print("Computer Wins!")
            break

        if " " not in board:
            print_board(board)
            print("Draw!")
            break


tic_tac_toe()
```
