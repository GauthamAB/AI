```python
import random


def calculate_cost(board):
    n = len(board)
    conflicts = 0

import random
import math


def calculate_cost(board):

    n = len(board)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):

            if board[i] == board[j]:
                conflicts += 1

            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing():

    n = 8

    board = [random.randint(0, n - 1) for i in range(n)]

    temperature = 1000
    cooling_rate = 0.99

    while temperature > 0.1:

        if calculate_cost(board) == 0:
            print("Solution Found!")
            break

        new_board = board[:]

        col = random.randint(0, n - 1)
        new_board[col] = random.randint(0, n - 1)

        old_cost = calculate_cost(board)
        new_cost = calculate_cost(new_board)

        if new_cost < old_cost:
            board = new_board

        else:
            probability = math.exp(
                (old_cost - new_cost) / temperature
            )

            if random.random() < probability:
                board = new_board

        temperature *= cooling_rate

    print("Final Board:")

    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

    print("Final Cost:", calculate_cost(board))


simulated_annealing()
```
