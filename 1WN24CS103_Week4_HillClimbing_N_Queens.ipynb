```python
import random


def calculate_cost(board):
    n = len(board)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):

            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def hill_climbing():

    n = 4
    board = list(range(n))
    random.shuffle(board)

    while True:

        current_cost = calculate_cost(board)

        print("Board:", board)
        print("Cost:", current_cost)

        if current_cost == 0:
            print("Solution Found!")
            break

        best_board = board[:]
        best_cost = current_cost

        for i in range(n):
            for j in range(i + 1, n):

                new_board = board[:]

                new_board[i], new_board[j] = (
                    new_board[j], new_board[i]
                )

                cost = calculate_cost(new_board)

                if cost < best_cost:
                    best_cost = cost
                    best_board = new_board

        if best_cost >= current_cost:
            print("Local Minimum Reached. Restarting...")
            board = list(range(n))
            random.shuffle(board)

        else:
            board = best_board


hill_climbing()
```
